```python
from vllm import LLM, SamplingParams
```

```python
# llm = LLM(model="facebook/opt-125m", enforce_eager=True)
llm = LLM(model="facebook/opt-125m")
```

```python
outputs = llm.generate(["what color is the sky?"], SamplingParams(temperature=0.8, max_tokens=50))
```

```python
print(outputs[0].outputs[0].text)
```

```python
from vllm import LLM, SamplingParams
```

```python
# llm = LLM(model="moonshotai/Kimi-K2-Instruct") # Too big for 1 GPU - 1 trillion parameter MoE model
llm = LLM(model="moonshotai/Kimi-K2-Instruct", tensor_parallel_size=6)
```

```python
outputs = llm.generate(["Hello!"], SamplingParams(temperature=0.7, max_tokens=256))
```

```python
print(outputs[0].outputs[0].text)
```
