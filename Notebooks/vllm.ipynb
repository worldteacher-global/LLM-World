```python
from vllm import LLM, SamplingParams
import os
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
```

```python
# llm = LLM(model="moonshotai/Kimi-K2-Instruct") # Too big for 1 GPU - 1 trillion parameter MoE model
# llm = LLM(model="moonshotai/Kimi-K2-Instruct", tensor_parallel_size=8)
# llm = LLM(model="moonshotai/Kimi-K2-Instruct",tensor_parallel_size=8,max_model_len=4096,gpu_memory_utilization=0.95,enforce_eager=True)
# llm = LLM(model="moonshotai/Kimi-K2-Instruct",tensor_parallel_size=8,max_model_len=4096,enforce_eager=True,cpu_offload_gb=20) # --mem=250G
llm = LLM(model="moonshotai/Kimi-K2-Instruct",
        tensor_parallel_size=8,
        max_model_len=4096,
        gpu_memory_utilization=0.95,
        enforce_eager=True,
        cpu_offload_gb=200)
```

```python
outputs = llm.generate(["Hello!"], SamplingParams(temperature=0.7, max_tokens=256))
```

```python
print(outputs[0].outputs[0].text)
```
